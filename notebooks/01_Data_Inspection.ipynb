```python
# Import required libraries

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Display settings
pd.set_option("display.max_columns", None)
pd.set_option("display.max_rows", 100)

# Visualization theme
sns.set_theme(style="whitegrid")

print("Libraries imported successfully!")
```

```python
# Inspect categorical columns

categorical_columns = df.select_dtypes(
    include="object"
).columns

for col in categorical_columns:
    print(f"\nColumn: {col}")
    print("Unique Values:", df[col].nunique())
    print(df[col].value_counts(dropna=False).head(10))
```

```python
# Statistical summary of numeric columns

df.describe()
```

```python
# Identify all duplicate Order IDs

duplicate_orders = df[
    df["Order_ID"].duplicated(keep=False)
].sort_values("Order_ID")

duplicate_orders[
    ["Order_ID", "Customer_ID", "Product_Name",
     "Quantity", "Sales_Amount", "Profit"]
].head(20)
```

```python
# Compare missing values with order status

status_missing_report = df.groupby(
    "Order_Status"
).agg(
    Total_Orders=("Order_ID", "count"),
    Missing_Delivery_Days=("Delivery_Days", lambda x: x.isna().sum()),
    Missing_Product_Rating=("Product_Rating", lambda x: x.isna().sum())
)

status_missing_report
```

```python
# Inspect date formats and invalid dates

date_check = pd.to_datetime(
    df["Order_Date"].astype("string").str.strip(),
    format="mixed",
    errors="coerce"
)

print("Total Date Records:", len(df))
print("Successfully Parsed:", date_check.notna().sum())
print("Invalid Dates:", date_check.isna().sum())

# Show invalid date examples
df.loc[
    date_check.isna(),
    ["Order_ID", "Order_Date"]
].head(20)
```

```python
# Create temporary numeric conversions for inspection

numeric_columns = [
    "Quantity",
    "Unit_Price",
    "Discount_Pct",
    "Sales_Amount"
]

for col in numeric_columns:

    temp = (
        df[col]
        .astype("string")
        .str.strip()
        .str.replace("₹", "", regex=False)
        .str.replace("units", "", regex=False)
        .str.replace(",", "", regex=False)
    )

    converted = pd.to_numeric(temp, errors="coerce")

    print(f"\nColumn: {col}")
    print("Conversion failures:", (converted.isna() & df[col].notna()).sum())
    print("Minimum value:", converted.min())
    print("Maximum value:", converted.max())
```

```python
# Create temporary numeric Sales column for comparison

sales_temp = (
    df["Sales_Amount"]
    .astype("string")
    .str.replace("₹", "", regex=False)
    .str.replace(",", "", regex=False)
    .str.strip()
)

df_audit = df.copy()
df_audit["Sales_Audit"] = pd.to_numeric(
    sales_temp, errors="coerce"
)

# Identify duplicate Order IDs
duplicate_report = (
    df_audit.groupby("Order_ID")
    .agg(
        Record_Count=("Order_ID", "size"),
        Unique_Customers=("Customer_ID", "nunique"),
        Unique_Products=("Product_ID", "nunique"),
        Unique_Sales_Values=("Sales_Audit", "nunique")
    )
    .query("Record_Count > 1")
    .sort_values("Record_Count", ascending=False)
)

duplicate_report
```

```python
conflicting_ids = duplicate_report[
    duplicate_report["Unique_Sales_Values"] > 1
].index

conflicting_records = df_audit[
    df_audit["Order_ID"].isin(conflicting_ids)
].sort_values("Order_ID")

conflicting_records[
    ["Order_ID", "Customer_ID", "Product_ID",
     "Quantity", "Sales_Audit", "Profit"]
].head(30)
```

```python
# Temporary numeric conversion

quantity_temp = pd.to_numeric(
    df["Quantity"].astype("string")
    .str.replace("units", "", regex=False)
    .str.strip(),
    errors="coerce"
)

discount_temp = pd.to_numeric(
    df["Discount_Pct"].astype("string")
    .str.replace("₹", "", regex=False)
    .str.strip(),
    errors="coerce"
)

print("Invalid Quantity Records:")
display(df.loc[quantity_temp <= 0, ["Order_ID", "Quantity"]])

print("Invalid Discount Records:")
display(df.loc[
    (discount_temp < 0) | (discount_temp > 100),
    ["Order_ID", "Discount_Pct"]
])
```

```python
# Create a separate audit DataFrame
# Original df will remain unchanged

financial_audit = df.copy()

# Convert financial columns temporarily into numeric values
for col in ["Sales_Amount", "Cost_Amount", "Profit"]:
    financial_audit[col] = pd.to_numeric(
        financial_audit[col]
        .astype("string")
        .str.replace("₹", "", regex=False)
        .str.replace(",", "", regex=False)
        .str.strip(),
        errors="coerce"
    )

# Calculate expected profit
financial_audit["Expected_Profit"] = (
    financial_audit["Sales_Amount"]
    - financial_audit["Cost_Amount"]
)

# Calculate difference
financial_audit["Profit_Difference"] = (
    financial_audit["Profit"]
    - financial_audit["Expected_Profit"]
)

# Flag inconsistent financial records
financial_audit["Financial_Check"] = np.where(
    np.isclose(
        financial_audit["Profit"],
        financial_audit["Expected_Profit"],
        atol=0.01,
        equal_nan=False
    ),
    "Valid",
    "Mismatch"
)

financial_audit["Financial_Check"].value_counts()
```

```python
financial_mismatches = financial_audit[
    financial_audit["Financial_Check"] == "Mismatch"
]

financial_mismatches[
    [
        "Order_ID",
        "Sales_Amount",
        "Cost_Amount",
        "Profit",
        "Expected_Profit",
        "Profit_Difference"
    ]
].head(25)
```

```python
# Inspect duplicate records with different sales values

conflict_financial_check = financial_audit[
    financial_audit["Order_ID"].isin(conflicting_ids)
].sort_values("Order_ID")

conflict_financial_check[
    [
        "Order_ID",
        "Sales_Amount",
        "Cost_Amount",
        "Profit",
        "Expected_Profit",
        "Financial_Check"
    ]
].head(30)
```

```python
# Identify financial mismatches linked to duplicate Order IDs

duplicate_order_ids = df["Order_ID"][
    df["Order_ID"].duplicated(keep=False)
].unique()

financial_mismatches["Duplicate_Order_Flag"] = (
    financial_mismatches["Order_ID"].isin(duplicate_order_ids)
)

financial_mismatches["Duplicate_Order_Flag"].value_counts()
```

```python
# Summarize financial mismatch differences

financial_mismatches["Profit_Difference"].value_counts().sort_index()
```
